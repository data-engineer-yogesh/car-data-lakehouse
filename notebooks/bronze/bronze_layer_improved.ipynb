Define Ingestion Configuration

```python
INGESTION_CONFIG = [
    {
        "source": "raw",
        "path": "/Volumes/cardatalakehouse_cat/landing/source_data/dim_car.csv",
        "table": "dim_car_info"
    },
    {
        "source": "raw",
        "path": "/Volumes/cardatalakehouse_cat/landing/source_data/dim_date.csv",
        "table": "dim_date_info"
    },
    {
        "source": "raw",
        "path": "/Volumes/cardatalakehouse_cat/landing/source_data/dim_dealer.csv",
        "table": "dim_dealer_info"
    },
    {
        "source": "raw",
        "path": "/Volumes/cardatalakehouse_cat/landing/source_data/dim_inventory.csv",
        "table": "dim_inventory_info"
    },
    {
        "source": "raw",
        "path": "/Volumes/cardatalakehouse_cat/landing/source_data/dim_review.csv",
        "table": "dim_review_info"
    },
    {
        "source": "raw",
        "path": "/Volumes/cardatalakehouse_cat/landing/source_data/fact_sales.csv",
        "table": "fact_sales_info"
    }
]
```

Ingest Files into Bronze Tables with loop

```python
for item in INGESTION_CONFIG:
    print(f"Ingesting {item['source']} → cardatalakehouse_cat.bronze.{item['table']}")

    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(item["path"])
    )

    (
        df.write
          .mode("overwrite")
          .format("delta")
          .saveAsTable(f"cardatalakehouse_cat.bronze.{item['table']}")
    )
```
